### Nhận xét nhóm House Age

Các biến liên quan đến thời gian xây dựng và sửa chữa có mối liên hệ
đáng chú ý với `LogSalePrice`. `YearBuilt` có correlation 0.5866,

0.5411.

Ngược lại, `YrSold` có correlation -0.0373, gần bằng 0. Điều này
cho thấy năm bán trong khoảng thời gian của dataset không có mối
liên hệ tuyến tính đáng kể với `LogSalePrice`.

Do `YearBuilt`, `YearRemodAdd` và `GarageYrBlt` biểu diễn các mốc
thời gian tuyệt đối, có thể chuyển đổi chúng thành các đặc trưng
tuổi tại thời điểm bán như `HouseAge`, `RemodAge` và `GarageAge`.
Các feature mới này sẽ được kiểm tra bằng thực nghiệm thay vì giả
định trước rằng chúng chắc chắn cải thiện mô hình.

```python
train["HouseAge"] = train["YrSold"] - train["YearBuilt"]
train["RemodAge"] = train["YrSold"] - train["YearRemodAdd"]

train["GarageAge"] = np.where(
    train["GarageYrBlt"].notna(),
    train["YrSold"] - train["GarageYrBlt"],
    np.nan
)
```

```python
# Các cột không được dùng làm feature
exclude_cols = [
    "Id",
    "SalePrice",
    "LogSalePrice"
]

# Toàn bộ feature số hiện có
numeric_features = train.select_dtypes(
    include=np.number
).columns.tolist()

numeric_features = [
    col for col in numeric_features
    if col not in exclude_cols
]

print("Số lượng numeric features:", len(numeric_features))
print("\nDanh sách:")
print(numeric_features)
```

```python
from sklearn.feature_selection import VarianceThreshold

# Chỉ dùng các numeric features đã xác định
X_numeric = train[numeric_features].copy()

# Điền missing tạm thời để tính variance
# Đây chỉ là bước khám phá feature, chưa phải bước huấn luyện model.
X_numeric_filled = X_numeric.fillna(
    X_numeric.median(numeric_only=True)
)

# Tìm các feature có variance <= 0.01
selector = VarianceThreshold(threshold=0.01)
selector.fit(X_numeric_filled)

selected_mask = selector.get_support()

low_variance_features = [
    feature
    for feature, keep in zip(numeric_features, selected_mask)
    if not keep
]

print("Số feature ban đầu:", len(numeric_features))
print("Số feature giữ lại:", selected_mask.sum())
print("Số feature variance thấp:", len(low_variance_features))

print("\nFeature variance thấp:")
print(low_variance_features)
```

```python
from sklearn.ensemble import RandomForestRegressor

# X và y cho bước phân tích feature importance
X_fs = train[numeric_features].copy()
y_fs = train["LogSalePrice"]

# Xử lý missing tạm thời
X_fs = X_fs.fillna(X_fs.median(numeric_only=True))

rf_fs = RandomForestRegressor(
    n_estimators=500,
    random_state=42,
    n_jobs=-1
)

rf_fs.fit(X_fs, y_fs)

feature_importance_fs = (
    pd.Series(
        rf_fs.feature_importances_,
        index=numeric_features
    )
    .sort_values(ascending=False)
)

display(
    feature_importance_fs.head(20).to_frame("Importance")
)
```

```python
# Tính ngưỡng feature importance cho EXP-04

importance_mean = feature_importance_fs.mean()
importance_threshold = 0.25 * importance_mean

selected_features_exp04 = (
    feature_importance_fs[
        feature_importance_fs >= importance_threshold
    ]
    .index
    .tolist()
)

print("Mean importance:", importance_mean)
print("Threshold:", importance_threshold)

print("\nSố feature được chọn:", len(selected_features_exp04))

print("\nFeatures được chọn:")
for feature in selected_features_exp04:
    print("-", feature)
```

```python
EXPERIMENTS["EXP-04"] = selected_features_exp04

print("EXP-04:")
print("Số feature:", len(EXPERIMENTS["EXP-04"]))
print("Features:")

for feature in EXPERIMENTS["EXP-04"]:
    print("-", feature)
```

```python
EXPERIMENTS["EXP-04"] = selected_features_exp04

print("EXP-04:")
print("Số feature:", len(EXPERIMENTS["EXP-04"]))
print("Features:")

for feature in EXPERIMENTS["EXP-04"]:
    print("-", feature)
```

```python
selection_report = pd.DataFrame({
    "feature": feature_importance_fs.index,
    "importance": feature_importance_fs.values,
    "selected_exp04": feature_importance_fs.index.isin(selected_features_exp04)
})

selection_report["threshold"] = importance_threshold

selection_report = selection_report.sort_values(
    "importance",
    ascending=False
)

selection_report.head(20)
```

5.4.2. Feature Engineering

Dựa trên kết quả phân tích đặc trưng, nhóm xây dựng thêm các feature tổng hợp nhằm biểu diễn rõ hơn những đặc điểm có ý nghĩa đối với giá nhà. Các feature được tạo gồm TotalSF, TotalBathrooms, TotalPorchSF, HouseAge, RemodAge và GarageAge.

TotalSF được xây dựng bằng cách tổng hợp diện tích basement và diện tích tầng 1, tầng 2. Feature này có correlation 0.777297 với log(SalePrice), cho thấy tổng diện tích sử dụng và basement có mối liên hệ đáng kể với giá bán.

TotalBathrooms tổng hợp số phòng tắm đầy đủ và quy đổi phòng tắm một phần theo trọng số 0.5. Feature này có correlation 0.673011 với log(SalePrice).

TotalPorchSF tổng hợp diện tích các khu vực ngoài trời gồm porch, deck và các loại diện tích tương ứng. Feature có correlation 0.398387, thấp hơn các feature tổng hợp về diện tích và phòng tắm nhưng vẫn cung cấp thêm thông tin về đặc điểm ngôi nhà.

Nhóm cũng xây dựng các feature tuổi nhà gồm HouseAge, RemodAge và GarageAge. Các feature này có correlation âm lần lượt là -0.587290, -0.568136 và -0.542110. Điều này cho thấy trong dữ liệu khảo sát, thời gian sử dụng càng lớn có xu hướng đi cùng mức log(SalePrice) thấp hơn.

Tuy nhiên, correlation chỉ phản ánh mối quan hệ tuyến tính giữa từng feature và biến mục tiêu, không đảm bảo feature mới sẽ cải thiện hiệu năng mô hình. Vì vậy, các feature này tiếp tục được kiểm chứng thông qua các thí nghiệm EXP-01, EXP-02 và EXP-03.
